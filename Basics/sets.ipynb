```python
# add elements
numbers = {1,2,3}
numbers.add(4)
print(numbers)
```

```python
#add mutliple elements
numbers = {1,2}
numbers.update([3,4,5])
print(numbers)
```

```python
#remove elements
numbers = {1,2,3}
numbers.remove(2)
print(numbers)
```

```python
numbers = {1,2,3}
numbers.discard(2)
print(numbers)
```

```python
#pop() for :- removing arbitrary element
numbers = {10,20,30}

x = numbers.pop()

print(x)
print(numbers)
```

```python
#clear()

numbers = {1,2,3}
numbers.clear()
print(numbers)
```

```python
#Membership ;- sets are efficient for checking membership
numbers = {10,20,30,40}
print(20 in numbers)
```

```python
#set operations
a = {1,2,3,4}
b ={3,4,5,6}
#union = all unique elements 
print(a|b)
a.union(b)
```

```python
#intersection - common elements
a = {1,2,3,4}
b ={3,4,5,6}
print(a&b)
a.intersection(b)
```

```python
#difference 
a = {1,2,3,4}
b ={3,4,5,6}
print(a - b)
print(b - a)
```

```python
#symmetric difference - print only that elements that exist in a and b , but not common
a = {1,2,3,4}
b ={3,4,5,6}
print(a^b)
```

```python
#subset / superset
a ={1,2}
b ={1,2,3,4}
print(a.issubset(b))
print(b.issuperset(a))
```

```python
#set from list - This is a quick way to identify unique values.

data =[1,2,2,3,4,4,5]

unique_data = set(data)

print(unique_data)
```

```python
#remove duplicates from list

data = [10,20,20,30,30,40]

unique = list(set(data))

print(unique)

#But remember: order may not be preserved
```
