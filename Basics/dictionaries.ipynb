```python
numbers = [1, 2, 2, 3, 3, 3]

frequency = {}

for num in numbers:

    if num in frequency:
        frequency[num] += 1

    else:
        frequency[num] = 1

print(frequency)
```

```python
frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num,0) +1
    
print(frequency)
```

```python
text = "banana"

frequency = {}

for char in text:
    frequency[char] = frequency.get(char,0)+1
print(frequency)
```

```python
#dictionary comprehension

squares = {}

for i in range(1,6):
    squares[i] = i **2
print(squares)
```

```python
squares = {i: i ** 2 for i in range(1, 6)}
print(squares)
```

```python
even_squares = {
    i: i ** 2
    for i in range(1, 11)
    if i % 2 == 0
}
print(even_squares)
```
